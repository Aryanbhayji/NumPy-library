1. np.dot() – Dot Product

It multiplies two arrays mathematically and combines their values through summation.

```python

import numpy as np

A = np.array([1, 2, 3])
B = np.array([4, 5, 6])

result = np.dot(A, B)

print("Dot Product:", result)
```

Where is it used in ML?

Imagine a Linear Regression model:

y = w_1x_1+w_2x_2+w_3x_3+b


Here, dot product calculates the weighted sum of input features.

This is one of the core operations in machine learning.

2. np.matmul() – Matrix Multiplication

It performs matrix multiplication between two arrays.

Unlike element-wise multiplication, matrix multiplication combines rows of one matrix with columns of another.

```python

import numpy as np

A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

result = np.matmul(A, B)

print(result)
```

Where is it used in ML?

Neural networks use matrix multiplication to calculate outputs from inputs and weights.

For example:

Z = XW+b


Where:
- X = Input features
- W = Model weights
- b = Bias
- Z = Calculated output

3. np.transpose() – Transpose Matrix

It converts rows into columns and columns into rows.

```python

import numpy as np

A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

result = np.transpose(A)

print("Original:")
print(A)

print("Transposed:")
print(result)

print("New Shape:", result.shape)
```

Where is it used in ML?
- Changing matrix dimensions.
- Preparing data for mathematical operations.
- Linear Regression calculations.
- Neural network computations.
A particularly important formula in Linear Regression is:

W=(X^TX)^{-1}X^Ty


Here, X^T means the transpose of the input matrix.

4. np.linalg.inv() – Matrix Inverse

It calculates the inverse of a square matrix, provided that the matrix is invertible.

```python

import numpy as np

A = np.array([
    [4, 7],
    [2, 6]
])

inverse = np.linalg.inv(A)

print("Inverse Matrix:")
print(inverse)

print("Verification:")
print(np.matmul(A, inverse))
```

Where is it used in ML?

One application is the closed-form solution of Linear Regression:

W=(X^TX)^{-1}X^Ty


However, directly calculating an inverse is often less numerically stable and efficient than using np.linalg.solve() or least-squares methods.

Important: An inverse does not exist if the matrix is singular.

5. Determinant – np.linalg.det()

A determinant is a single numerical value calculated from a square matrix.

```python

import numpy as np

A = np.array([
    [2, 1],
    [1, 1]
])

determinant = np.linalg.det(A)

print("Determinant:", determinant)
```

Why does this matter?

A determinant of zero indicates that the square matrix is singular.

In linear regression, redundant features can make the normal-equation matrix non-invertible.

ML applications:
- Detecting redundant information.
- Understanding multicollinearity.
- Diagnosing problems with regression models.

6. Matrix Rank – np.linalg.matrix_rank()


ML application: Feature selection and dimensionality reduction

Suppose your dataset has two columns that contain exactly the same information.

Your dataset technically has two features, but only one independent feature.

```python

import numpy as np

X = np.array([
    [1, 2],
    [2, 4],
    [3, 6],
    [4, 8]
])

rank = np.linalg.matrix_rank(X)

print("Number of Features:", X.shape[1])
print("Matrix Rank:", rank)
```

Why ML needs this:
- Identifying redundant features.
- Understanding dataset dimensions.
- Diagnosing multicollinearity.
- Understanding PCA and dimensionality reduction.
