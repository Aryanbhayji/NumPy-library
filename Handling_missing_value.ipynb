```python
import numpy as np  
```

```python
#.np.isnan(array)

arr = np.array([1,2,np.nan,5,np.nan,6])
print(np.isnan(arr))
```

```python
#np.nan_to_num(array , nan=value ) --. default is 0

arr = np.array([1,2,np.nan,5,np.nan,6])

clear_arr = np.nan_to_num(arr)
print(clear_arr)

```

```python
#np.isinf(array) ---> return boolean if value is infinity or - infinity

arr = np.array([1,2,np.inf,5,-np.inf,6])
print(np.isinf(arr))


clear_arr = np.nan_to_num(arr , posinf=1000,neginf=-1000)
print(clear_arr)
```
