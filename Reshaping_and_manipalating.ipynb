```python
import numpy as np
```

```python
#reshaping ---> reshape(row,coloum),specify new shape id dim match 

arr= np.array([1,2,3,4,5,6,])
reshape_arr= arr.reshape(2,3)
print(reshape_arr)
```

```python
#[.ravel--> view , affect orignal array
#.flatten--> copy, doesn't affect orignal array ]  ---> to convert multidimension array into 1d

arr= np.array([[1,2,3],[4,5,6]]) 
print(arr.ravel())
print(arr.flatten())
```
