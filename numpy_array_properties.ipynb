```python
import numpy as np
```

```python
#shape --> numbers of row and coloum in maxtrices

a= np.array([[1,2,3],
            [4,5,6]])

print(a.shape)
```

```python
#size : gives the total number of elements in array

a= np.array([[1,2,3],
            [4,5,6]])

print(a.size)
```

```python
#ndim : number of dimension

a= np.array([[1,2,3],
            [4,5,6]])

b= np.array([1,2,3])

print(a.ndim )
print(b.ndim)
```

```python

#dtype: data type of elements

b= np.array([1,2.5,3])
print(b.dtype)
```

```python
#conversion of type --> astype()

arr = np.array([1,2,3,4,5])
int_arr = arr.astype(int)

print(int_arr)
print(int_arr.dtype)
```
