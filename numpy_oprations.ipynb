```python
import numpy as np
```

```python
arr = np.array([10,20,30])

print(arr * 2)
print(arr ** 2)
print(arr + 10)
```

```python
#sum 
arr = np.array([10,20,30])

print(np.sum(arr))

#mean

print(np.mean(arr))

#min

print(np.min(arr))

#max 

print(np.max(arr))

#standarded deviation  

print(np.std(arr))

#cal variance

print(np.var(arr))
```
