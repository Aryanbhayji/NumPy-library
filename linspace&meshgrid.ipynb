```python
#linspace(start, stop, num=50, endpoint=True, retstep=False, dtype=None, axis=0)

import numpy as np

x= np.linspace(0, 10, 10)
y=np.linspace(0, 10, 10)

print(x)
print(y)
```

```python
#meshgrid(x,y)-->converts 1D arrays to 2D arrays

x= np.linspace(0, 10, 10)
y=np.linspace(0, 10, 10)

X, Y = np.meshgrid(x, y)

print(X)
print(Y)
```
