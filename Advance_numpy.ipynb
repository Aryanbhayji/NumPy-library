```python
import numpy as np
```

```python
#insert(arr,index,value,axis=None ), axis=0 --> row wise , axis=1 --> coloum wise


arr = np.array([10,20,30,40,50,60])

new_arr = np.insert(arr,2,100)
print(new_arr)

arr_2d = np.array([[1,2,3],[4,5,6]])
new_arr_2d = np.insert(arr_2d,2,[6,8,5], axis=0)
print(new_arr_2d)
```

```python
#append (arr,value)
arr = np.array([1,2,3])
print(np.append(arr,[4,5,6]))
```

```python
#np.concatenate((array1,array2),axis =0)
#axis 0 > vertically stack
#axis 1 > horizontally stack

arr1= np.array([1,2,3])
arr2= np.array([5,6,7])

new_arr= np.concatenate((arr1,arr2))
print(new_arr)
```

```python
#np.delete(arr,index,axis= None)
#flatten

arr = np.array([10,20,30,40,50,60])
new_arr = np.delete(arr,0)
print(new_arr)

arr_2d = np.array([[1,2,3],[4,5,6]])
new_arr_2d = np.delete(arr_2d,0,axis=0)
print(new_arr_2d)  
```

```python
#stacking ---> vstack = row wise
#hstack = coloumm wise


arr1= np.array([1,2,3])
arr2= np.array([5,6,7])

print(np.vstack((arr1,arr2)))
print(np.hstack((arr1,arr2)))
```

```python
#np.vsplit ,np.hsplit

arr = np.array([10,20,30,40,50,60])
print(np.split(arr , 2))
```

```python
#🔢 Fancy Indexing

import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print(arr[[0, 2, 4]])
```

```python
# 🔍 Boolean Indexing

arr = np.array([10, 20, 30, 40, 50])

print(arr[arr > 25])
```

```python
#📡 Broadcasting

arr = np.array([[1, 2, 3],
                [4, 5, 6]])

print(arr + 10)
```
