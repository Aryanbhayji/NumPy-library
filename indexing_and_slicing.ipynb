```python
import numpy as np
```

```python
#array[] --> id-array
#array[row, coloum] --> 2d-array

arr = np.array([2,3,4,5])

print(arr[0])#frist element
print(arr[-1])#last element
```

```python
#slicing ---> arr[start:end:step]

arr = np.array([2,3,4,5,6,7,8])

print(arr[1:-1:2])
print(arr[::-1])
```

```python
#fancy indexting


arr = np.array([2,3,4,5,6,7,8])
print(arr[[0,3,6]])

```

```python
#boolean masking


arr = np.array([2,3,4,5,6,7,8])
print(arr[arr<5])
```
